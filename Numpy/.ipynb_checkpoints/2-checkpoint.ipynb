### Data type (dtype)

```python
import numpy as np 
```

```python
data = np.random.randn(3, 5)
dt = data.dtype
dt.itemsize
```

```python
dt.name
```

```python
dt.type
```

```python
dt.type is np.float64
```

```python
np.float64
```

```python
# change a data type with 'astype'
data32 = data.astype(np.float32)
print(data.dtype)
print(data32.dtype)
```

### Arithmetic operations

```python
arr = np.arange(1, 21)
```

```python
arr2 = arr.reshape(5, 4)
arr2
```

```python
arr2 ** 2
```

```python
arr2 / 5
```

```python
1 / arr2
```

```python
arr[5] = 200
arr.reshape(5, 4) > arr2
```

```python
rng = np.random.default_rng()
rng.normal(size=(5, 10))
```

```python
arr2[1:4]
```

```python
arr2[1, :3]
```

```python
arr2[1, 1:3]
```
