### Geometric representation

```python
# create a vector
v = np.array([-2,3])

plt.arrow(0,0,v[0],v[1], head_width=.3, width=.1, color='k')
plt.plot(0,0, 'ko', markerfacecolor='k', markersize=8)

plt.plot([-4,4],[0,0], '--', color='k', zorder=-1)
plt.plot([0,0],[-4,4], '--', color='k', zorder=-1)

plt.axis('square')
plt.axis([-4,4,-4,4])
plt.xlabel('$v_0$')
plt.ylabel('$v_1$')
plt.title('Vector v in standard position')
plt.show()
```

```python
startposition = [
    [-1,-1],
    [0,0],
    [2,-1]
]
for a in startposition:
    plt.arrow(a[0],a[1],v[0],v[1], head_width=.3, width=.1,color='k')
    plt.plot(a[0],a[1], 'ko', markerfacecolor='k', markersize=4)

plt.plot([-5,5],[0,0], '--', color='k',zorder=-1)
plt.plot([0,0],[-5,5], '--', color='k', zorder=-1)

plt.axis('square')
plt.axis([-5,5,-5,5])
plt.xlabel('$v_0$')
plt.ylabel('$v_1$')
plt.title('Vector $\\mathbf{v}$ in various locations')
plt.show()
```

### exercise 2:
    Write an algorithm that computes the norm of a vector by translating

```python
# this code belowe compute the norme of an vector, like np.linalg.norm(array)
from math import sqrt

def normeVector(a):
    return sqrt(np.sum(a ** 2))
    
v1 = np.array([0,0,1]) # unit vector
v2 = np.array([ [10,9,8],[1,2,3],[-4,6,8] ]) # non unit

print("Norme of v1:", normeVector(v1))
print("Norme of v1 with numpy:", np.linalg.norm(v1))

print("Norme of v2:", normeVector(v2))
print("Norme of v2 with numpy:", np.linalg.norm(v2))
```

### exercise 3:
    Create a Python function that will take a vector as input and output a unit vector in
    the same direction. What happens when you input the zeros vector?

```python
def unitVector(v):
    return v / np.linalg.norm(v)

u = np.array([0,1,0]) # unit vector
print(unitVector(u))

w = np.array([0,0,4]) # non unit 
print(unitVector(w))

z = np.array([4,9,8])
print(unitVector(z))

print(np.zeros((4, 4)))
```

### exercise 4:

Write a for loop to transpose a row vector into a column vector without using a
built-in function or method such as np.transpose() or v.T.

```python
a1 = np.array([ [1,2,3] ])
a2 = np.empty((a1.size, 1), dtype=int)

for i in np.arange(a1.size):
    a2[i, 0] = a1[0, i]

a2
```

### exercise 5
Here is an interesting fact: you can compute the squared norm of a vector as the dot
product of that vector with itself.

```python
a3 = np.array([1,2,3,4,5])

dot_pro = np.dot(a3, a3)
norm_squa = normeVector(a3) ** 2

print(dot_pro)
print(norm_squa)
```

### exercise 6:
Write code to demonstrate that the dot product is commutative.
Commutative means that a × b = b × a, which, for the vector dot product, means that aTb = bTa.

```python
a4 = np.array([1,2,3,4])
a5 = np.array([5,6,7,8])

e = np.array([ [1,2],
               [3,4] ])
f = np.array([ [5,6],
               [7,8] ])
left = np.dot(a4, a5)
rigth = np.dot(a5, a4)
              
print(f"left: {left} and rigth: {rigth}")
print(np.dot(e.T, f))
print(np.dot(f.T, e))
```
