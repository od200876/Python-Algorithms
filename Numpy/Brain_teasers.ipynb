### Given the following data type and data set; sort according to 'height'

```python
camelot_dtype = [('name', 'S10'), ('height', float), ('age', int)]
camelot_value = [('Arthur', 1.8, 41), ('Lancelot', 1.9, 38), ('Galahad', 1.6, 39)]
```

```python
A_dtype = np.asarray(camelot_value, dtype=camelot_dtype)
A_dtype
```

```python
for value in np.sort(A_dtype, order='height'):
    print(value)
```
